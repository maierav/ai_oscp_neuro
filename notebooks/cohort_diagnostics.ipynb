# Cohort diagnostics — inclusion flow, cross-modality per-animal, sensorimotor equivalence

Three cohort-level figures, all built from committed tables in `data/` (no DANDI streaming):

1. **Inclusion flow** (CONSORT-style): sessions → CCF → QC units → responsiveness → final analysis set, per paradigm (Neuropixels) and per modality (cross-scale feature-oddball).
2. **Cross-modality per-animal DvI**: one point per animal, matched areas (VISp/VISl), Neuropixels vs mesoscope, with responsiveness-gate sensitivity. SLAP2 excluded (no control-referenced DvI).
3. **Sensorimotor equivalence + confound**: a pre-specified SESOI TOST and a Bayes factor on the closed−open null, plus the structural block-order confound.

All figures use a colorblind-safe (Okabe-Ito) palette and are saved as PNG + PDF + SVG.

```python
fig, ov = cfc.make_inclusion_flow()
print("inclusion_flow overlaps:", len(ov)); fig
```
